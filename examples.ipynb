```python
import re
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.transfermarkt.com.br/erling-haaland/profil/spieler/418560"
player_id = url.split('/')[-1] # esta pegando o ultimo item da lista seperando por /
```

```python
headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
} #usually required to acsses websites
```

```python
response = requests.get(url, headers=headers)
```

```python
response.status_code # checking if everything is allright
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
player_name = soup.select_one('h1[class="data-header__headline-wrapper"]').text.split('\n')[-1].strip()
# selecting the object in the webpage, only giving the text. (.text)
# \n dividing eand creating a list.
# [-1] lat part only
# strip cleaning the white space
```

```python
player_number = soup.select_one('span[class="data-header__shirt-number"]').text.replace('#','').strip()
# found the tag span on the webpage, only giving the text. (.text)
# replace, for replacing the # in front
# strip for cleaning the white space
```

```python
player_contrct = re.search("Contrato até: .*__content\">(.*?)</span>", str(soup)).group(1)

display(player_contrct)
```
