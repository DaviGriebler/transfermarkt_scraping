```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://www.transfermarkt.com.br/premier-league/spieltag/wettbewerb/GB1/saison_id/2025/spieltag/1"

headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
}

response = requests.get(url, headers=headers)

response.status_code
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
roundSoup = soup.find_all('tr', {'class':'no-border spieltagsansicht-aktionen'})

display(roundSoup)
```

```python
print(roundSoup[0].find('a').get('href'))
print(roundSoup[0].find('a').get('title'))
print(roundSoup[0].find('td', {'class':'zentriert no-border-links'}).string)
print(roundSoup[0].find('td', {'class':'zentriert hauptlink'}).string)
```

```python
tabela = []

for i, row in enumerate(roundSoup):
    temp = []
    temp.append(row.find('a').get('href'))
    temp.append(row.find('a').get('title'))

    info2 = row.find('td', {'class':'zentriert no-border-links'}).string
    if info2 == '\xa0': temp.append(row.find('td', {'class':'zentriert no-border-rechts'}).string)
    else: temp.append(row.find('td', {'class':'zentriert no-border-links'}).string)

    try: temp.append(row.find('td', {'class':'zentriert hauptlink'}).string) 
    except AttributeError: continue
    
    tabela.append(temp)

display(tabela)
print(len(tabela))
```

```python
df = pd.DataFrame(tabela)
df.columns = ['player_link','player_name','minute_label','scoreboard']
display(df)
```
