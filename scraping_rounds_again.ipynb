```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://www.transfermarkt.com.br/premier-league/spieltag/wettbewerb/GB1/saison_id/2025/spieltag/1"

headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
}

response = requests.get(url, headers=headers)

response.status_code
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
all_information = soup.find_all('table', {'style':'border-top: 0 !important;'})

display(all_information)
display(len(all_information))
```

```python
home_team = all_information[9].find('td', {'class':'rechts hauptlink no-border-rechts hide-for-small spieltagsansicht-vereinsname'}).find('a').get('title')
away_team = all_information[9].find('td', {'class':'hauptlink zentriert no-border-rechts no-border-links hide-for-small spieltagsansicht-wappen'}).find('a').get('title')
final_score = all_information[9].find('span', {'class':'matchresult finished'}).string

adicional_info = all_information[9].find_all('td', {'class':'zentriert no-border'})
day = adicional_info[0].find('a').string
ref = adicional_info[1].find('a').string
attendance = re.search("span>([\s\S]*?)<", str(adicional_info[2])).group(1)

display(home_team)
display(away_team)
display(final_score)
display(day.strip())
display(ref)
display(attendance.strip())
```

```python
round_1 = []
for row in all_information:
    temp = []

    home_team = row.find('td', {'class':'rechts hauptlink no-border-rechts hide-for-small spieltagsansicht-vereinsname'}).find('a').get('title')
    away_team = row.find('td', {'class':'hauptlink zentriert no-border-rechts no-border-links hide-for-small spieltagsansicht-wappen'}).find('a').get('title')
    final_score = row.find('span', {'class':'matchresult finished'}).string

    temp.append(home_team)
    temp.append(final_score)
    temp.append(away_team)

    adicional_info = row.find_all('td', {'class':'zentriert no-border'})

    for i, item in enumerate(adicional_info):
        if i == 2:
            attendance = re.search(r"span>([\s\S]*?)<", str(item)).group(1)
            temp.append(attendance.strip())
        else:
            day_ref = item.find('a').string
            temp.append(day_ref.strip())

    round_1.append(temp)

display(round_1)
```

```python
df_round_1 = pd.DataFrame(round_1)
df_round_1.columns = ['home_team','score','away_team','date','referee','attendance']
display(df_round_1)
```
