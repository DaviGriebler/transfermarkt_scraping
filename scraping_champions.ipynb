```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://www.transfermarkt.com.br/premier-league/erfolge/wettbewerb/GB1"

headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
}

response = requests.get(url, headers=headers)

response.status_code
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
team_champSoup = soup.find_all('td', {'class':'hauptlink no-border-links'})
manager_champSoup = soup.find_all('td', {'class':'rechts'})

display(manager_champSoup)
```

```python
team_champSoup[0].string
team_champSoup[0].find('a').get('href')
```

```python
team_champ = []
clock = 0

for row in team_champSoup:
    if clock == 20: break # last 20 years
    temp = []
    temp.append(row.find('a').get('href'))
    temp.append(row.string)
    team_champ.append(temp)
    clock += 1

df_team_champ = pd.DataFrame(team_champ)

display(df_team_champ)
```

```python
manager_champSoup[0].string
manager_champSoup[0].find('a').get('id')
```

```python
manager_champ = []
clock = 0

for row in manager_champSoup:
    if clock == 20: break # last 20 years
    temp = []
    temp.append(row.find('a').get('href'))
    temp.append(row.find('a').get('id'))
    temp.append(row.string)
    manager_champ.append(temp)
    clock += 1

df_manager_champ = pd.DataFrame(manager_champ)

display(df_manager_champ)
```

```python
df_champions = pd.concat([df_team_champ,df_manager_champ], axis=1)
df_champions.columns = ['season_link', 'team_name', 'manager_link', 'manager_id', 'manager_name']
display(df_champions)
```
