```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://www.transfermarkt.com.br/premier-league/startseite/wettbewerb/GB1/plus/?saison_id=2025"

headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
}

response = requests.get(url, headers=headers)

response.status_code
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
teamSoup = soup.find_all('td', {'class':'rechts'})

display(teamSoup)
```

```python
display(teamSoup[3])

teamSoup[3]

display(teamSoup[3].string)

if 'href' in str(teamSoup[3]):
    print('yo')
```

```python
team_value = []

for i, row in enumerate(teamSoup):
    temp = []
    if "href" in str(row):
        temp.append(re.search('title="(.*)">€', str(row)).group(1))
        temp.append(row.string)
        team_value.append(temp)
    else: continue

df = pd.DataFrame(team_value)
df.columns = ['Clube','Value']
display(df)
```

```python
estrangeiro = soup.find_all('td', {'class':'zentriert'})
del estrangeiro[:3]
display(estrangeiro)
```

```python
#title="(.*)"><img

clock = 0
list_plantel = []
temp1=[]
for row in estrangeiro:
    if clock == 0:
        temp1.append(re.search('title="(.*)"><img', str(row)).group(1))
    if clock == 1:
        temp1.append(row.string)
    if clock == 3:
        temp1.append(row.string)
        clock = -1
        list_plantel.append(temp1)
        temp1=[]
    clock+=1
    if len(list_plantel) == 20:
        break

df_plantel = pd.DataFrame(list_plantel)
df_plantel.columns = ['Clube', 'Plantel', 'Estrangeiros']
display(df_plantel)
```
