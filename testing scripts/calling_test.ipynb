```python
from transfermarkt_script import Transfermarkt
import pandas as pd
```

```python
headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
}

premier_data = Transfermarkt('premier-league', headers)
```

```python
all_matches = []
for i,season in enumerate(range(2024,2026)):
    temp = premier_data.get_match(season)
    if i == 0: all_matches += temp
    else: all_matches += temp[1:]

display(all_matches)
```

```python
df = pd.DataFrame(all_matches[1:], columns=all_matches[0])

display(df)
```
