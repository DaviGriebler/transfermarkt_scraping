```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://www.transfermarkt.com.br/premier-league/tabelle/wettbewerb/GB1/saison_id/2025#google_vignette"

headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
}

response = requests.get(url, headers=headers)

response.status_code
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
body = soup.find_all('td', {'class':'rechts hauptlink'})

display(body)
```

```python
# teamSoup = soup.find_all('td', class_='zentriert')

teamSoup = soup.find_all('td', {'class':'zentriert'})

team = []

for i, row in enumerate(teamSoup):
    try: 
        team.append(re.search('img alt="(.*)" class="tiny_wappen"', str(row)).group(1))
    except AttributeError: continue

del team[-3:]

display(team)
```

```python
statSoup = soup.find_all('td', {'class':'zentriert'})
#class="zentriert">(.*)<

stat = []
temp = []
c = -1

for i, row in enumerate(statSoup):
    try: 
        temp.append(re.search('class="zentriert">(.*)<', str(row)).group(1))
        c += 1
    except AttributeError: continue
    if c == 6:
        stat.append(temp)
        temp = []
        c = -1

del stat[-1:]

display(stat)
```

```python
scoreBoard = []

for num_stat, actual_stat in enumerate(stat):
    temp = []
    temp.append(team[num_stat])
    for index_stat in actual_stat:
        temp.append(index_stat)
    scoreBoard.append(temp)

df = pd.DataFrame(scoreBoard)
df.columns = ['Clube','M','W','D','L','Goals','AG','Pts']
display(df)
```
