```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://www.transfermarkt.com.br/premier-league/tabelle/wettbewerb/GB1/saison_id/2025#google_vignette"

headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
}

response = requests.get(url, headers=headers)

response.status_code
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
tableSoup = soup.find_all('td', {'class' : 'zentriert'})

display(tableSoup)
```

```python
display(tableSoup[0].find('a').get('href'))
display(tableSoup[0].find('a').get('title'))
display(tableSoup[1].string)
```

```python
table = []
temp = []
clock = 0

for row in tableSoup:
    if clock == 7: 
        clock = 0
        table.append(temp)
        temp = []
    if clock == 0:
        try: 
            temp.append(row.find('a').get('href'))
            temp.append(row.find('a').get('title'))
        except: continue
    else: temp.append(row.string)
    clock += 1
del table[-1]

df_table = pd.DataFrame(table)
df_table.columns = ['team_link', 'team_name', 'played', 'win', 'draw', 'loss', 'goals', 'goal_dif']
display(df_table)
```
