```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) Gecko/20100101 Firefox/150.0"
}
```

```python
final_placement = []

for n_season in range(1992,2026):

    url = f"https://www.transfermarkt.com.br/premier-league/tabelle/wettbewerb/GB1/saison_id/{n_season}"
    response = requests.get(url, headers=headers)
    response.status_code
    soup = BeautifulSoup(response.content, "html.parser")

    all_info = soup.find_all('tbody')
    options = all_info[0].find_all('tr')
    info = all_info[1].find_all('tr')

    season_key = f'PL-{n_season}'

    print(f'{n_season}, ', end="")

    for i,row in enumerate(info):
        temp = []

        temp.append(season_key)

        position = i+1
        team = row.find('a').get('title')
        data_info = row.find_all('td', {'class':'zentriert'})

        temp.append(position)
        temp.append(team)

        for i, item in enumerate(data_info):
            if i == 0: continue
            temp.append(item.string)
        
        final_placement.append(temp)

df_final_placement = pd.DataFrame(final_placement)
df_final_placement.columns = ['season_id', 'pos','team_name','played','wins','draws','losses','goals','goal_dif','points']
display(df_final_placement)
```
